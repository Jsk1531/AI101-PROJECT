```python
import pandas as pd
df=pd.read_csv(r"Used_cars.csv")
print("Dataset shape (rows, columns):", df.shape)
df.head()
```

```python

missing_values = df.isnull().sum().sort_values(ascending=False)

print("Missing values per column:")
print(missing_values[missing_values > 0])
```

```python
from IPython.display import display

print("Dataset dimensions:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

print("\nFirst five rows:")
display(df.head())

print("\nData types and missing-value overview:")
df.info()

print("\nSummary of numeric columns:")
display(df.describe())
```

```python
df["price"]=pd.to_numeric(df["price"].str.replace('$','',regex=False).str.replace(',','',regex=False),errors='coerce')
df["milage"]=pd.to_numeric(df["milage"].str.replace('mi.','',regex=False).str.replace(',','',regex=False),errors='coerce')
display(df)
```

```python
import matplotlib.pyplot as plt
import numpy as np

numeric_columns = df.select_dtypes(include=np.number).columns.tolist()
print("Numeric columns:", numeric_columns)
print(numeric_columns)
column_to_plot = numeric_columns[-1]

plt.figure(figsize=(9,5))
plt.hist(df[column_to_plot].dropna(), bins=20,range=(0,150000), edgecolor="black")
plt.title(f"Distribution of {column_to_plot}")
plt.xlabel(column_to_plot)

plt.grid(alpha=0.25)
plt.show()
```

```python
feature_for_plot = numeric_columns[1]
target_for_plot = numeric_columns[-1]  # Check that this is actually the price column!
df_normal=df[df[target_for_plot]<=150000]
plt.figure(figsize=(8, 5))
plt.scatter(df_normal[feature_for_plot], df_normal[target_for_plot], alpha=0.65)
plt.title(f"{target_for_plot} vs {feature_for_plot}")
plt.xlabel(feature_for_plot)
plt.ylabel(target_for_plot)
plt.grid(alpha=0.25)
plt.show()
```

```python
# Average price of the 10 most common car brands

brand_df = df.dropna(subset=["brand", "price"]).copy()

brand_df = brand_df[brand_df["price"] <= 150000]

top_brands = brand_df["brand"].value_counts().head(10).index

average_price_by_brand = (
    brand_df[brand_df["brand"].isin(top_brands)]
    .groupby("brand")["price"]
    .mean()
    .sort_values()
)

print("Average price by brand:")
print(average_price_by_brand.round(2))

plt.figure(figsize=(10, 6))
average_price_by_brand.plot(kind="bar", color="steelblue", edgecolor="black")

plt.title("Average Used-Car Price by Brand")
plt.xlabel("Brand")
plt.ylabel("Average Price (USD)")
plt.xticks(rotation=45, ha="right")
plt.grid(axis="y", alpha=0.25)

plt.show()
```
